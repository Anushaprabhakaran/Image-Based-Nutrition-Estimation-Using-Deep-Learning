```python
import os
import pandas as pd
import numpy as np
from PIL import Image
from sklearn.model_selection import train_test_split
import torch
from torch import nn, optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models

# Paths
CSV_FILE = "/content/labels_subset.csv"      # Your filtered labels CSV
IMG_DIR = "/content/drive/MyDrive/nutrition5k_resized"          # Resized images folder

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Dataset class
class NutritionDataset(Dataset):
    def __init__(self, csv_file, img_dir, transform=None):
        self.data = pd.read_csv(csv_file)
        self.img_dir = img_dir
        self.transform = transform
        self.labels = self.data[['total_calories', 'total_fat', 'total_carb', 'total_protein']].values
        self.filenames = self.data['dish_id'].values

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        img_name = self.filenames[idx] + "_rgb.png"
        img_path = os.path.join(self.img_dir, img_name)
        image = Image.open(img_path).convert('RGB')

        if self.transform:
            image = self.transform(image)

        label = torch.tensor(self.labels[idx], dtype=torch.float32)
        return image, label, self.filenames[idx]

# Transformations
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# Load dataset
dataset = NutritionDataset(CSV_FILE, IMG_DIR, transform=transform)

# Train-test split (80%-20%)
train_size = int(0.8 * len(dataset))
test_size = len(dataset) - train_size
train_dataset, test_dataset = torch.utils.data.random_split(dataset, [train_size, test_size])

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

# Model - Pretrained ResNet50 with output layer for 4 nutrition values
model = models.resnet50(pretrained=True)
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 4)
model = model.to(device)

# Loss and optimizer
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training function
def train_model(model, dataloader, criterion, optimizer, epochs=10):
    model.train()
    for epoch in range(epochs):
        running_loss = 0.0
        for inputs, labels, _ in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            running_loss += loss.item() * inputs.size(0)

        epoch_loss = running_loss / len(dataloader.dataset)
        print(f"Epoch {epoch+1}/{epochs}, Loss: {epoch_loss:.4f}")

# Evaluation function
def evaluate_model(model, dataloader, criterion):
    model.eval()
    running_loss = 0.0
    with torch.no_grad():
        for inputs, labels, _ in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            running_loss += loss.item() * inputs.size(0)

    total_loss = running_loss / len(dataloader.dataset)
    print(f"Test Loss (MSE): {total_loss:.4f}")

# Train the model
train_model(model, train_loader, criterion, optimizer, epochs=10)

# Evaluate the model
evaluate_model(model, test_loader, criterion)

# Predict a single image (example)
def predict_single_image(model, img_filename):
    model.eval()
    img_path = os.path.join(IMG_DIR, img_filename)
    image = Image.open(img_path).convert("RGB")
    image = transform(image).unsqueeze(0).to(device)

    with torch.no_grad():
        pred = model(image)
    pred_np = np.round(pred.cpu().numpy(), 2)[0]

    dish_id = img_filename.replace("_rgb.png", "")
    print(f"Prediction for {dish_id}:")
    print(f"  Calories: {pred_np[0]}")
    print(f"  Fat:      {pred_np[1]}")
    print(f"  Carbs:    {pred_np[2]}")
    print(f"  Protein:  {pred_np[3]}")

# Example prediction
predict_single_image(model, "dish_1556572657_rgb.png")
```

```python
import os
import pandas as pd
import numpy as np
from PIL import Image
import torch
from torch import nn, optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

# Paths
CSV_FILE = "/content/labels_subset.csv"
IMG_DIR = "/content/drive/MyDrive/nutrition5k_resized"
SAVE_DIR = "/content/drive/MyDrive/nutrition_checkpoints"

# Create save folder if it doesn't exist
os.makedirs(SAVE_DIR, exist_ok=True)

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device, flush=True)

# Dataset class
class NutritionDataset(Dataset):
    def __init__(self, csv_file, img_dir, transform=None):
        self.data = pd.read_csv(csv_file)
        self.img_dir = img_dir
        self.transform = transform
        self.labels = self.data[['total_calories', 'total_fat', 'total_carb', 'total_protein']].values
        self.filenames = self.data['dish_id'].values

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        img_name = self.filenames[idx] + "_rgb.png"
        img_path = os.path.join(self.img_dir, img_name)
        image = Image.open(img_path).convert('RGB')
        if self.transform:
            image = self.transform(image)
        label = torch.tensor(self.labels[idx], dtype=torch.float32)
        return image, label, self.filenames[idx]

# Transformations
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# Load dataset
dataset = NutritionDataset(CSV_FILE, IMG_DIR, transform=transform)
print(f"Dataset size: {len(dataset)}", flush=True)
sample_img, sample_label, sample_id = dataset[0]
print(f"Sample loaded: {sample_id}, label: {sample_label}", flush=True)

# Train-test split
train_size = int(0.8 * len(dataset))
test_size = len(dataset) - train_size
train_dataset, test_dataset = torch.utils.data.random_split(dataset, [train_size, test_size])
print(f"Train size: {train_size}, Test size: {test_size}", flush=True)

# Reduced batch size to 16 to avoid memory crash
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True, num_workers=2, pin_memory=True)
test_loader  = DataLoader(test_dataset,  batch_size=16, shuffle=False, num_workers=2, pin_memory=True)

# Model
model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 4)
model = model.to(device)
print("Model loaded successfully!", flush=True)

# Check if a saved checkpoint exists and resume from it
checkpoint_path = None
for i in range(10, 0, -1):
    path = f"{SAVE_DIR}/nutrition_model_epoch{i}.pth"
    if os.path.exists(path):
        checkpoint_path = path
        start_epoch = i
        break

if checkpoint_path:
    model.load_state_dict(torch.load(checkpoint_path))
    print(f"✅ Resumed from saved checkpoint: epoch {start_epoch}", flush=True)
else:
    start_epoch = 0
    print("Starting fresh training", flush=True)

# Loss and optimizer
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training
def train_model(model, dataloader, criterion, optimizer, total_epochs=10, start_epoch=0):
    model.train()
    for epoch in range(start_epoch, total_epochs):
        running_loss = 0.0
        for i, (inputs, labels, _) in enumerate(dataloader):
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * inputs.size(0)
            if i % 10 == 0:
                print(f"  Epoch {epoch+1}, Batch {i}/{len(dataloader)}", flush=True)

        epoch_loss = running_loss / len(dataloader.dataset)
        print(f"✅ Epoch {epoch+1}/{total_epochs}, Loss: {epoch_loss:.4f}", flush=True)

        # Save after every epoch to Drive
        save_path = f"{SAVE_DIR}/nutrition_model_epoch{epoch+1}.pth"
        torch.save(model.state_dict(), save_path)
        print(f"💾 Saved: {save_path}", flush=True)

# Evaluation
def evaluate_model(model, dataloader, criterion):
    model.eval()
    running_loss = 0.0
    with torch.no_grad():
        for inputs, labels, _ in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            running_loss += loss.item() * inputs.size(0)
    total_loss = running_loss / len(dataloader.dataset)
    print(f"✅ Test Loss (MSE): {total_loss:.4f}", flush=True)

# Predict single image
def predict_single_image(model, img_filename):
    model.eval()
    img_path = os.path.join(IMG_DIR, img_filename)
    image = Image.open(img_path).convert("RGB")
    image = transform(image).unsqueeze(0).to(device)
    with torch.no_grad():
        pred = model(image)
    pred_np = np.round(pred.cpu().numpy(), 2)[0]
    dish_id = img_filename.replace("_rgb.png", "")
    print(f"\nPrediction for {dish_id}:")
    print(f"  Calories: {pred_np[0]}")
    print(f"  Fat:      {pred_np[1]}")
    print(f"  Carbs:    {pred_np[2]}")
    print(f"  Protein:  {pred_np[3]}")

# Run training (auto resumes if checkpoint exists)
train_model(model, train_loader, criterion, optimizer, total_epochs=10, start_epoch=start_epoch)

# Evaluate
evaluate_model(model, test_loader, criterion)

# Predict example
predict_single_image(model, "dish_1561662216_rgb.png")
```
